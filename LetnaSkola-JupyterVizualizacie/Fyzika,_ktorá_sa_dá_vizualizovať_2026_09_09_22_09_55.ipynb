```sage
# Ukážka grafu funkcie (sinusová vlna s tlmením)

# Definícia funkcie f1(x) = sin(x)/x (funkcia s tlmením)
f1(x) = 1/x*sin(x)
farba1 = "green"

# Horná a dolná obálka
f2(x) = 1/x
f3(x) = -1/x
farba23 = "red"

# Nastavenie rozsahov osí
xmin, xmax = -30, 30
ymin, ymax = -0.3, 1

# Vykreslenie hlavnej funkcie s legendou
graf1 = plot(f1, (x, xmin, xmax), ymin=ymin, ymax=ymax, 
             color=farba1, 
             legend_label=r"$\dfrac{1}{x}~\sin x$")

# Vykreslenie obmedzujúcich funkcií (bez legendy)
graf2 = plot(f2, (x, xmin, xmax), ymin=ymin, ymax=ymax, color=farba23)
graf3 = plot(f3, (x, xmin, xmax), ymin=ymin, ymax=ymax, color=farba23)

# Kombinácia grafov (aktuálne len hlavná funkcia)
vysledny_graf = graf1  # + graf2 + graf3
# Ďalšie grafy sa dajú zobraziť "pripočítaním" ďalších grafov

# Zobrazenie výsledného grafu s rámom a mriežkou
vysledny_graf.show(frame=True, 
                   gridlines=True, 
                   title="Ukážka grafu v SageMath", 
                   axes_labels=("$x$","$y$"))
```

## **3. Od Newtonovej gravitácie k potenciálu**

### **Newtonov gravitačný zákon**

Isaac Newton zistil, že každé dve telesá s hmotnosťami $m_1$ a $m_2$ vo vzdialenosti $r$ na seba pôsobia gravitačnou silou:

$$F = \kappa \frac{m_1 m_2}{r^2}$$

kde $\kappa = 6{,}674 \times 10^{-11} \, \mathrm{N \cdot m}^2/\text{kg}^2$ je gravitačná konštanta, $m_1, m_2$ sú hmotnosti telies a $r$ je vzájomná vzdialenosť telies.

**Problém:** 

Sila je **vektor** (má smer aj veľkosť). Pri gravitačnom pôsobení viacerých telies je obtiažne vektorovým súčtom získať výslednú silu.

Napríklad dve sily $\vec{F_1}$, $\vec{F_2}$ s veľkosťou $10~\mathrm{N}$ môžu v súčte $\vec{F} = \vec{F_1} + \vec{F_2}$ dávať $\vec{F}$ s veľkosťou napr. $20~\mathrm{N}$ ($\vec{F_1}$, $\vec{F_2}$ s rovnakým smerom), $0~\mathrm{N}$ (opačne orientované  $\vec{F_1}$, $\vec{F_2}$) alebo čokolvek medzi tým (napr. $\sqrt{200}~\mathrm{N}$, ak sú $\vec{F_1}$, $\vec{F_2}$ na seba kolmé).

### **Gravitačný potenciál – elegantnejší prístup**

Namiesto síl (vektorov) môžeme použiť **gravitačný potenciál** $\phi$:

$$\phi(r) = -\kappa \frac{M}{r}$$

Potenciál má niekoľko výhod:

* Je to **skalár** (len číslo, nie vektor)
* Potenciály od viacerých telies sa jednoducho **sčítajú** ako klasické čísla
* Zo zmeny potenciálu (skaláru) vieme získať späť silu (vektor)

> **Fyzikálny význam:**
> * Gravitačný potenciál je **gravitačná potenciálna energia objektu s jednotkovou hmotnosťou $1~\mathrm{kg}$** v gravitačnom poli objektu s hmotnosťou $M$ vo vzdialenosti $r$. Tým pádom je jednotkou potenciálu $1 ~\mathrm{J/kg}$ (joule na kilogram), po úprave $1 ~\mathrm{m^2/s^2}$.
> * Táto energia je vždy záporná, lebo ju uvažujeme vzhľadom na **nulovú hladinu v nekonečne**. 
> * Je to podobné, ako keď potenciálna energia guľôčky umiestnenej pod stolom je záporná vzhľadom na povrch stola (lebo “výška nad stolom” je záporná). Je len potrebné si predstaviť, že nech je guľôčka kdekolvek vo vesmíre, stále je “pod stolom”, ktorého povrch je nekonečne ďaleko.

### **Skalárne pole**

Keď každému bodu v priestore $(x, y)$ priradíme číselnú hodnotu (skalár), vytvoríme **skalárne pole**. Skalárne pole je vlastne špeciálny názov pre niečo, s čím sa stretávame aj v bežnom živote:

* **Mapa nadmorských výšok:** každému miestu určenému geografickou polohou $(\mathrm{geografická~šírka}, \mathrm{geografická~dĺžka})$ priraďujeme číslo - nadmorskú výšku.
* **Teplotná mapa:** podobným štýlom, každému miestu priraďujeme číselnú hodnotu - teplotu (napr. na snímku z termokamery)
* **Mapa tlaku vzduchu**

Gravitačný potenciál je tiež len číslo. Ak by sme chceli vytvoriť **gravitačnú mapu**, môžeme polohe v priestore $(x, y)$ priradiť hodnotu gravitačného potenciálu $\phi$.

### **Úloha 1: Zadanie gravitačného potenciálu pre Zem**

Najprv vytvoríme funkciu, ktorá vypočíta gravitačný potenciál v ľubovoľnom bode $(x, y)$ spôsobený Zemou.

$$\phi_{Zem}(x,y) = -\kappa \frac{M_{Zem}}{r(x,y)}$$

kde:

* $\kappa$ je gravitačná konštanta
* $M_{Zem}$ je hmotnosť telesa (Zem)
* $r(x,y)$ je vzdialenosť od začiatku súradnicovej sústavy/miesta, kde sa nachádza teleso (Zem)


**Prirodzené jednotky pre Zem:**

Pre zjednodušenie numerických výpočtov a vizualizácie používame prirodzené jednotky, kde kladieme:

* **Hmotnosť Zeme:** $M_{Zem} = 1$ (jednotková hmotnosť)
* **Gravitačná konštanta:** $\kappa = 1$


> ***Pozn.:*** V skutočnosti pracujeme s extrémne veľkými a malými číslami, napr. $M_{Zem} = 5{,}972 \times 10^{24} ~ \mathrm{kg}$, $\kappa = 6{,}674 \times 10^{-11} ~ \mathrm{m^3\cdot kg^{-1} \cdot s^{-2}}$, ktoré prirodzené jednotky normalizujú na hodnoty okolo 1.

> **Otázka:** Ako môžeme určiť vzdialenosť od Zeme $r$ v závislosti od polohy telesa danej súradnicami $(x,y)$?

```sage
# Definícia premenných a gravitačného potenciálu Zeme
x, y = var('x y')

# Prirodzené jednotky
M_Zem = 1
kappa = 1

# Vzdialenosť od Zeme (v strede súradnicovej sústavy)
r(x, y) = sqrt(x^2 + y^2)

# Gravitačný potenciál Zeme
phi_Zem(x, y) = -kappa * M_Zem / r(x, y)
show(phi_Zem)
```

### **Úloha 2: Gravitačný potenciál Mesiaca**

Teraz pridáme Mesiac. Potrebujeme upraviť funkciu tak, aby počítala vzdialenosť od pozície Mesiaca.

**Prirodzené jednotky pre Mesiac:**

* Hmotnosť Mesiaca: $M_{\text{Mesiac}} = 0{,}0123 ~ M_{Zem}$ (relatívna hmotnosť voči Zemi)
* Vzdialenosť Zem-Mesiac: $d = 1$ (jednotková vzdialenosť)
* Budeme počítať v sústave, kde Zem je v bode $(0, 0)$ a Mesiac v bode $(d, 0)$

> ***Pozn.:*** Opäť, uvažujeme prirodzené jednotky, lebo v skutočnosti $M_{Mesiac} = 7{,}347 \times 10^{22} ~ \mathrm{kg}$ ; $d = 3{,}844 \times 10^{11} ~ \mathrm{m}$

> **Tip:** Zrejme bude potrebné určiť vzdialenosť od Mesiaca, ktorý ale nie je v počiatku súradnicovej sústavy v bode $(0,0)$. Dá sa ukázať, že ak máme bod $O = (a,b)$, vzdialenosť od tohto bodu bude $r_O (x,y)= \sqrt{(x-a)^2 + (y-b)^2}$ . Ako sa to dá uplatniť pre Mesiac $M=(0,d)$?

```sage
# Definícia gravitačného potenciálu Mesiaca v sústave Zem-Mesiac

# Parametre Mesiaca v prirodzených jednotkách
M_Mesiac = 0.0123  # hmotnosť Mesiaca v jednotkách hmotnosti Zeme
d = 1              # vzdialenosť Zem-Mesiac

# Pozícia Mesiaca v súradnicovej sústave (Zem je v (0,0))
a, b = d, 0        # Mesiac je v bode (d, 0)

# Vzdialenosť od Mesiaca
r_Mesiac(x, y) = sqrt((x - a)^2 + (y - b)^2)

# Gravitačný potenciál Mesiaca
phi_Mesiac(x, y) = -kappa * M_Mesiac / r_Mesiac(x, y)
show(phi_Mesiac)
```

### **Úloha 3: Spoločný gravitačný potenciál**

Tu prichádza tá krásna vec – potenciály sa jednoducho **sčítajú**!

Celkový potenciál v bode $(x, y)$ je:

$$\phi_{\text{celk.}}(x, y) = \phi_{\text{Zem}}(x, y) + \phi_{\text{Mesiac}}(x, y)$$

```sage
# Celkový gravitačný potenciál (Zem + Mesiac)
phi_celkovy(x,y) = phi_Zem(x,y) + phi_Mesiac(x,y)

# Zobrazenie celkového potenciálu
show(phi_celkovy(x,y))
```

## **6. Lagrangeove body - miesta rovnováhy v kozme**

V predchádzajúcich úlohách sme vizualizovali gravitačný potenciál systému Zem-Mesiac. Videli sme "krajinu" s dvomi hlbokými "studňami" (Zem a Mesiac). Ale existujú aj špeciálne body v priestore okolo Zeme a Mesiaca, ktoré sa nazývajú sa **Lagrangeove body** a sú kľúčové pre kozmonautiku

Predstav si, že si astronaut a chceš sa dostať na miesto vo vesmíre, kde by si "visel" medzi Zemou a Mesiacom bez toho, aby ťa niektoré z týchto telies priťahovalo viac. Takéto miesta v priestore nazývame **Lagrangeove body** (alebo tiež libračné body) a je ich presne **päť**.

**Čo sú Lagrangeove body?**

Lagrangeove body sú špeciálne miesta v priestore okolo dvoch nebeských telies (napríklad Zem a Mesiac), **kde sa gravitačné sily oboch telies a odstredivá sila presne vyrovnávajú**. Ak by si sa dostal do takého bodu, mohol by si tam zostať a otáčať sa spolu so Zemou a Mesiacom, bez toho aby si musel používať motory na udržanie pozície.

<p style="text-align: center;"><img src="https://i.postimg.cc/sxQnv6tv/lagrange-points.png" alt="image" width="100%" /></p>

<p style="text-align: center;"><em>Obr: Lagrangeove body L1 až L5 v systéme Jupiter-Slnko. V L4 a L5 sú "uväznené" malé asteroidy Trójania a Gréci (zdroj: <a href="https://science.mainguet.org/spacetime/lagrange/lagrange.htm" target="_blank">science.mainguet.org</a>)</em></p>

**Prečo je to užitočné?**

V praxi sú Lagrangeove body ideálne miesta na umiestnenie:

* Vesmírnych observatórií (napr. James Webb teleskop je v bode L2 systému Slnko-Zem)
* Komunikačných satelitov
* Kozmických staníc pre medziplanetárne misie

**Kde sa nachádzajú?**

Existuje päť Lagrangeových bodov, označených L1 až L5:

* **L1** - nachádza sa medzi Zemou a Mesiacom. Odtiaľto je dobrý výhľad na obidve telesá.
* **L2** - je za Mesiacom (ďalej od Zeme).
* **L3** - je na opačnej strane Zeme od Mesiaca. Zem ti bráni vo výhľade na Mesiac.
* **L4** - je 60° pred Mesiacom na jeho obežnej dráhe (tvorí rovnostranný trojuholník so Zemou a Mesiacom)
* **L5** - je 60° za Mesiacom na jeho obežnej dráhe (taktiež rovnostranný trojuholník)

> **Zaujímavosť:** Francúzsky matematik Joseph-Louis Lagrange tieto body objavil už v roku 1772 pri štúdiu tzv. "problému troch telies"

Aby sme mohli zistiť polohu Lagrangeových bodov, potrebujeme dokončiť naše modelovanie gravitačného poľa Zeme a Mesiaca.

1. Potrebujeme **zohľadniť odstredivú silu** (zatiaľ sme to neurobili) $\Rightarrow$ tomu zodpovedá **potenciál odstredivej sily**, ktorý je potrebné pripočítať k nášmu potenciálu
2. **Potenciál** - skalárnu veličinu, potrebuje premeniť **na silu** - vektorovú veličinu $\Rightarrow$ to dokážeme urobiť veľmi elegantne pomocou **gradientu skalárneho poľa** (bude ozrejmené neskôr)
3. Potrebujeme určiť polohu Lagrangeových bodov na základe podmienky **nulovej výslednej sily**

### **Úloha 7: Zohľadnenie odstredivej sily**

Aby sme pochopili Lagrangeove body, musíme uvažovať v **rotujúcom referenčnom systéme** - teda v systéme súradníc, ktorý sa otáča spolu so Zemou a Mesiacom. V tomto systéme Zem a Mesiac stoja na mieste, ale pôsobí na nás **odstredivá sila**.

**Čo je odstredivá sila?**

Keď sa niečo otáča, pociťujeme silu, ktorá nás tlačí smerom von od stredu otáčania. Ak si už niekedy sedel na kolotoči, poznáš ten pocit, keď ťa "vymršťuje" von. To je odstredivá sila - nie je to skutočná sila, ale efekt rotácie referenčného systému.

**Matematický popis odstredivej sily:**

Ak sa systém Zem-Mesiac otáča s uhlovou rýchlosťou $\omega$, potom odstredivá sila pôsobiaca na teleso vo vzdialenosti $r$ je:

$$F_{odstr} = m \; \omega^2 \; r $$

kde $r$ je vzdialenosť od stredu rotácie - ťažiska systému Zem-Mesiac. V prípade Zeme a Mesiaca je ich spoločné ťažisko  pod povrchom Zeme, a teda môžeme situáciu trochu zjednodušiť a uvažovať polohu tohto ťažiska presne v strede Zeme.

> ***Pozn.:*** To by ale nemusela byť pravda v iných systémoch. Napr. v systéme Pluto-Cháron (mesiac Pluta) je ich spoločné ťažisko mimo objemu planétky Pluto. Cháron tak de facto neobieha okolo Pluta, ale spoločne rotujú okolo ťažiska, ktoré sa nachádza v priestore medzi nimi.

**Potenciál odstredivej sily:**

Každú konzervatívnu silu môžeme vyjadriť pomocou potenciálu. Pre odstredivú silu je potenciál:

$$\phi_{odstr} = -\frac{1}{2}\omega^2 r^2$$

Pre vzdialenosť $r$ od stredu otáčania bude potrebné použiť opäť podobný trik s Pytagorovou vetou a uhlovú rýchlosť pre jednoduchosť môžeme položiť $\omega = 1$. Pri perióde $T= 27{,}32 ~\mathrm{d} = 2{,}36 \cdot 10^6 ~\mathrm{s}$ by skutočná uhlová rýchlosť bola $\omega= 2{,}66 \cdot 10^{-6}~\mathrm{rad \cdot s^{-1}}$ (extrémne malá hodnota).

Zarátaním $\phi_{odstr.}$ k celkovému potenciálu dostaneme **Rocheho potenciál**:

$$\phi_{Roche} = \phi_{Zem} + \phi_{Mesiac} + \phi_{odstr.}$$

Skúsme teda zadať Rocheho potenciál a zobraziť si ho podobne, ako v predošlej časti.

```sage
# Uhlová rýchlosť
omega = 1

# Potenciál odstredivej sily
phi_odstr(x, y) = -1/2 * omega^2 *  r(x, y)^2

phi_Roche (x, y) = phi_Mesiac(x,y) + phi_Zem(x,y) + phi_odstr(x,y)
show(phi_Roche)
```

```sage
# Vykreslenie gravitačného potenciálu pozdĺž x-ovej osy (y=0) s odrezaním singulárnych bodov

# Rozsah zobrazenia pozdĺž x-ovej osy
xmin_profil, xmax_profil = -1.5, 2.5

# 2D graf potenciálu "z profilu"
graf_profil = plot(
    phi_Roche(x,0), 
    (x, xmin_profil, xmax_profil),
    color="red",
    thickness=2,
    ymin=hranicna_hodnota,
    legend_label=r"$\phi_{\text{celkový}}(x,0)$"
)

# Pridanie zvislých čiar pre pozíciu Zeme a Mesiaca
ciara_Zem = line([(-0.02, -10), (-0.02, 0)], color="green", linestyle="--", legend_label="Zem")
ciara_Mesiac = line([(d-0.02, -10), (d-0.02, 0)], color="gray", linestyle="--", legend_label="Mesiac")

# Kombinácia grafov
vysledny_profil = graf_profil + ciara_Zem + ciara_Mesiac

# Zobrazenie s popiskami
vysledny_profil.show(
    frame=True,
    gridlines=True,
    title="Rocheho potenciál pozdĺž x-ovej osy (Zem-Mesiac)",
    axes_labels=["$x$", "$\\phi(x,0)$"],
    ymin=-10, ymax=0
)
```

### **Úloha 8: Od potenciálu k sile**

Doteraz sme pracovali s **gravitačným potenciálom** $\phi$ - skalárnou veličinou, ktorá popisuje gravitačné pole. No v praxi nás často zaujíma **sila**, ktorá pôsobí na telesá v tomto poli. Práve preto zavedieme pojem **intenzita gravitačného poľa**.

#### **Čo je intenzita gravitačného poľa?**

**Intenzita gravitačného poľa** $\vec{E}$ je **gravitačná sila pôsobiaca na teleso jednotkovej hmotnosti $1~\mathrm{kg}$**.

Možno si si všimol, že jednotka intenzity gravitačného poľa je:
$$[\vec{E}] = \frac{\mathrm{N}}{\mathrm{kg}} = \frac{\mathrm{m}}{\mathrm{s}^2}$$

Táto jednotka $\mathrm{N/kg}$ ti možno pripomína niečo zo základnej školy! Keď sa učíš o gravitácii na Zemi, hovoríme o **gravitačnom zrýchlení** $g = 9{,}81~\mathrm{m/s^2}$, ktoré môžeme zapísať aj ako $g = 9{,}81~\mathrm{N/kg}$.

**Intenzita gravitačného poľa je vlastne gravitačné zrýchlenie v danom mieste priestoru.** Udáva nám, akým zrýchlením by sa pohybovalo teleso, keby sme ho v tom mieste "pustili".

#### **Vzťah medzi potenciálom a intenzitou: Gradient**

Ako získame intenzitu $\vec{E}$ z potenciálu $\phi$? Pomocou matematickej operácie nazvanej **gradient**:

$$\vec{E} = -\nabla \phi$$

**Čo je gradient?**

Gradient (označovaný symbolom $\nabla$ "nabla", alebo len prosto $\operatorname{grad}$) je **diferenciálny operátor**, ktorý zo skalárneho poľa (čísla v každom bode priestoru) vytvorí **vektorové pole** (vektor v každom bode priestoru). 

Intuitívne: **gradient ukazuje smer najväčšieho rastu funkcie** (v našom prípade potenciálu). Záporné znamienko v $\vec{E} = -\nabla \phi$ znamená, že gravitačná sila smeruje **v smere najväčšieho poklesu potenciálu** - teleso "padá" tam, kde je potenciál najnižší.

> **Pre záujemcov – matematická definícia:**
> 
> Gradient je vektor zložený z parciálnych derivácií:
> $$\nabla \phi = \left(\dfrac{\partial \phi}{\partial x}, \dfrac{\partial \phi}{\partial y}\right)$$
> 
> Parciálna derivácia $\dfrac{\partial \phi}{\partial x}$ meria, ako rýchlo sa potenciál mení v smere osi $x$ (pri konštantnom $y$).

#### **Prečo nás zaujíma intenzita pri hľadaní Lagrangeových bodov?**

Lagrangeove body sme definovali ako miesta, kde sa **gravitačné sily vyrovnávajú**. Podľa **druhého Newtonovho zákona**:

$$\vec{F} = m \vec{a}$$

Ak je **intenzita gravitačného poľa (gravitačné zrýchlenie) nulová** $\vec{E} = \vec{0}$, potom je aj **sila nulová** $\vec{F} = \vec{0}$.

**Naša úloha sa teda mení:**

> Nájdi miesta v priestore, kde je **intenzita gravitačného poľa nulová**: $\vec{E}(x,y) = \vec{0}$

To je oveľa konkrétnejšia úloha, ktorú dokážeme vyriešiť numericky aj graficky!

#### **SageMath kód: Výpočet intenzity z Rocheho potenciálu**

Teraz použi `.gradient()` funkciu v SageMath na získanie intenzity gravitačného poľa z Rocheho potenciálu:

```sage
# Výpočet intenzity gravitačného poľa z Rocheho potenciálu
# Intenzita E = -gradient(phi_Roche)

# Gradient potenciálu (vektor so zložkami Ex, Ey)
gradient_phi_Roche = phi_Roche(x, y).gradient()

# Intenzita gravitačného poľa (opačný smer gradientu)
E_x(x, y) = -gradient_phi_Roche[0]  # zložka v smere x
E_y(x, y) = -gradient_phi_Roche[1]  # zložka v smere y

# Zobrazenie vzorcov pre intenzitu
show(E_x)
show(E_y)
```

### **Úloha 9: Vizualizácia intenzity gravitačného poľa**

#### **Vektorové pole**

Keďže intenzita $\vec{E}$ je **vektor** v každom bode priestoru, môžeme ju zobraziť ako **vektorové pole** - v každom bode nakreslíme šípku, ktorá ukazuje smer a veľkosť gravitačnej sily.

```sage
# Vizualizácia intenzity gravitačného poľa ako vektorového poľa

# Rozsah zobrazenia
xmin, xmax = -1.5, 2.5
ymin, ymax = -1.5, 1.5

# vektor (Ex, Ey)
E = vector([E_x(x, y), E_y(x, y)])    

# Vektorové pole intenzity
vektorove_pole = plot_vector_field(
    E,
    (x, xmin, xmax),
    (y, ymin, ymax),
    plot_points=20,             # hustota šípok
    color='red'
)

# Pridanie pozícií Zeme a Mesiaca
bod_zem = points([(0, 0)], color='cyan', size=80, zorder=5, legend_label='Zem')
bod_mesiac = points([(d, 0)], color='grey', size=60, zorder=5, legend_label='Mesiac')

# Zobrazenie
vysledok = vektorove_pole + bod_zem + bod_mesiac
vysledok.show(
    frame=True,
    title="Vektorové pole intenzity gravitačného poľa",
    axes_labels=["$x$", "$y$"]
)
```

#### **Normalizované vektorové pole**

Ako sme mohli vidieť, pri vizualizácii vektorového poľa môže nastať problém: **šípky majú veľmi rôzne dĺžky**. V miestach blízko Zeme alebo Mesiaca je intenzita veľká (dlhé šípky), zatiaľ čo ďalej od týchto telies je intenzita malá (krátke šípky, ktoré sú takmer neviditeľné).

Riešením je **normalizácia** vektorového poľa. Pri normalizácii každý vektor **zachováva svoj smer**, ale všetky vektory majú **rovnakú dĺžku** (typicky 1). Tým sa lepšie vidí **smer** gravitačnej sily v celom priestore.

**Matematicky:**

Pre vektor $\vec{E} = (E_x, E_y)$ je normalizovaný vektor:

$$\vec{E}_{\text{norm}} = \frac{1}{|\vec{E}|} \cdot \vec{E} = \frac{1}{\sqrt{E_x^2 + E_y^2}} \cdot (E_x, E_y)$$

Na normalizovanom vektorovom poli je oveľa lepšie vidieť **smer gravitačnej sily** v každom bode priestoru. Sledovaním šípok môžeme určiť trajektórie , ktorými by sa teleso pohybovalo. **Lagrangeove body** sa dajú identifikovať ako miesta, kde sa šípky "stretávajú" z rôznych smerov alebo kde niet šípok vôbec

```sage
# Vizualizácia intenzity gravitačného poľa ako NORMALIZOVANÉHO vektorového poľa

E_norm = E/E.norm()

# Normalizované vektorové pole
vektorove_pole_norm = plot_vector_field(
    E_norm,
    (x, xmin, xmax),
    (y, ymin, ymax),
    plot_points=20,        
    color='red',
    zorder=10
)


# Zobrazenie
vysledok = vektorove_pole_norm + bod_zem + bod_mesiac
vysledok.show(
    frame=True,
    title="Normalizované vektorové pole intenzity gravitačného poľa",
    axes_labels=["$x$", "$y$"],
)
```

#### **Veľkosť intenzity gravitačného poľa**

Okrem smeru nás zaujíma aj **veľkosť** intenzity $|\vec{E}|$. Podľa **Pytagorovej vety** pre zložky vektora:

$$|\vec{E}(x,y)| = \sqrt{E_x^2 + E_y^2}$$

```sage
# Výpočet veľkosti intenzity gravitačného poľa
E_velkost(x, y) = sqrt(E_x(x, y)^2 + E_y(x, y)^2)

# Zobrazenie vzorca pre veľkosť intenzity
show(E_velkost(x, y))
```

#### **Vizualizácia veľkosti intenzity**

Podobne ako pri potenciáli, zobrazíme veľkosť intenzity troma spôsobmi:

**a) Graf pozdĺž osi $x$ (profil)**

```sage
# Vykreslenie veľkosti intenzity pozdĺž x-ovej osy (y=0)

xmin_profil, xmax_profil = -1.5, 2.5

# Graf profilu intenzity
graf_profil_E = plot(
    E_velkost(x, 0), 
    (x, xmin_profil, xmax_profil),
    color="purple",
    thickness=2,
    legend_label="veľkosť intezity E",
    ymin=0,
    ymax=3
)

# Pridanie zvislých čiar pre pozíciu Zeme a Mesiaca
ciara_Zem = line([(0, 0), (0, 3)], color="green", linestyle="--", thickness=4, legend_label="Zem")
ciara_Mesiac = line([(d, 0), (d, 3)], color="grey", linestyle="--", thickness=4, legend_label="Mesiac")

# Kombinácia grafov
vysledny_profil_E = ciara_Zem + ciara_Mesiac +  graf_profil_E 

# Zobrazenie
vysledny_profil_E.show(
    frame=True,
    gridlines=True,
    title="Veľkosť intenzity gravitačného poľa pozdĺž x-ovej osy",
    axes_labels=["$x$", r"$|E|~(x,0)$"]
)
```

**Čo vidíme?** 

Na grafe by mali byť viditeľné **tri body, kde intenzita klesá na nulu** - to sú Lagrangeove body **L1, L2 a L3** na spojnici Zem-Mesiac!

**b) 3D vizualizácia veľkosti intenzity**

```sage
# 3D graf veľkosti intenzity gravitačného poľa

# Definícia rozsahu
xmin, xmax = -1.5, 2.5
ymin, ymax = -1.5, 1.5

# 3D graf s odrezaním príliš veľkých hodnôt
hranica_E = 3
def E_velkost_odrezana(x_val, y_val):
    if x_val==0 and y_val==0:
        return hranica_E
    elif E_velkost(x=x_val, y=y_val) < hranica_E:
        return E_velkost(x=x_val, y=y_val) 

    else:
        return hranica_E

graf_3d_E = plot3d(
    E_velkost_odrezana,
    (x, xmin, xmax),
    (y, ymin, ymax),
    plot_points=120,
    color='magenta',
    mesh=False,
    alpha=0.4
)

# Zobrazenie
graf_3d_E
```

**c) Teplotná mapa veľkosti intenzity**

```sage
# Teplotná mapa veľkosti intenzity gravitačného poľa (logaritmická škála)

# Definícia rozsahu
xmin, xmax = -1.5, 2.5
ymin, ymax = -1.5, 1.5

# Teplotná mapa
teplotna_mapa_E1 = contour_plot(
    log(E_velkost,10),
    (x, xmin, xmax),
    (y, ymin, ymax),
    cmap='viridis',
    contours=25,
    colorbar=True,
    colorbar_label=r"$\log(|E|)$",
    plot_points=200,
    fill=True,
    zorder=-20
)

# Pridanie pozícií Zeme a Mesiaca
bod_zem = points([(0, 0)], color='cyan', size=80, zorder=5, legend_label='Zem')
bod_mesiac = points([(d, 0)], color='lightgrey', size=60, zorder=5, legend_label='Mesiac')

# Kombinácia
vysledna_mapa_E = teplotna_mapa_E1 + bod_zem + bod_mesiac 

# Zobrazenie
(vysledna_mapa_E + vektorove_pole_norm).show(
    frame=True,
    title="Teplotná mapa veľkosti intenzity gravitačného poľa",
    axes_labels=["$x$", "$y$"],
    figsize=10
)
```

### **Úloha 10: Hľadanie Lagrangeových bodov L1, L2, L3**

Teraz vieme, že Lagrangeove body L1, L2 a L3 ležia na **spojnici Zem-Mesiac** (os $x$, kde $y=0$). Hľadáme miesta, kde:

$$|\vec{E}|(x, 0) = 0$$

Tieto body nájdeme **numericky** pomocou funkcie `find_root`:

```sage
# Hľadanie koreňov (núl) veľkosti intenzity pozdĺž x-ovej osy

# Definícia funkcie pre hľadanie koreňov
E_na_osi_x(x) = E_velkost(x=x, y=0)
show(E_na_osi_x)
```

```sage
# Hľadanie L1 (medzi Zemou a Mesiacom)
x_L1 = find_root(E_na_osi_x(x) - 0.001, 0.7, 0.95)

# Hľadanie L2 (za Mesiacom)
x_L2 = find_root(E_na_osi_x(x) - 0.001, 1.05, 1.3)

# Hľadanie L3 (za Zemou)
x_L3 = find_root(E_na_osi_x(x) - 0.001, -1.2, -0.8)

x_L1, x_L2, x_L3
```

```sage
# Súradnice Lagrangeových bodov
L1 = [x_L1, 0]
L2 = [x_L2, 0]
L3 = [x_L3, 0]

show(L1)
show(L2)
show(L3)
```

```sage
# Vizualizácia na mape
body_L123 = points([L1, L2, L3], color='red', marker='x', size=100, zorder=7, legend_label='L1, L2, L3')

vysledna_mapa_s_L123 = vysledna_mapa_E + body_L123
vysledna_mapa_s_L123.show(
    frame=True,
    title="Lagrangeove body L1, L2, L3",
    axes_labels=["$x$", "$y$"],
    figsize=10
)
```

### **Úloha 11: Lagrangeove body L4 a L5 – Geometrická úloha**

Body **L4 a L5** neležia na spojnici Zem-Mesiac, ale vo **vrcholoch rovnostranných trojuholníkov**, ktorých základňou je spojnica Zem-Mesiac.

**Geometrická úloha:**

Zem je v bode $(0, 0)$, Mesiac v bode $(d, 0)$. Hľadáme body $L_4$ a $L_5$ tak, aby trojuholníky Zem-Mesiac-$L_4$ a Zem-Mesiac-$L_5$ boli **rovnostranné**.

Pre rovnostranný trojuholník so základňou dĺžky $d$ je výška:
$$h = \frac{\sqrt{3}}{2} d$$

Stred základne (stred medzi Zemou a Mesiacom) je v bode $\left(\frac{d}{2}, 0\right)$.

Súradnice bodov L4 a L5:
$$L_4 = \left(\frac{d}{2}, \frac{\sqrt{3}}{2} d\right)$$
$$L_5 = \left(\frac{d}{2}, -\frac{\sqrt{3}}{2} d\right)$$

```sage
# Výpočet súradníc L4 a L5

# Vzdialenosť Zem-Mesiac
d = 1

# Súradnice L4 a L5 (vrcholy rovnostranných trojuholníkov)
x_L4_L5 = d / 2
y_L4 = sqrt(3) / 2 * d
y_L5 = -sqrt(3) / 2 * d

L4 = [x_L4_L5, y_L4]
L5 = [x_L4_L5, y_L5]

show( L4)
show(L5)
```

```sage
# Vizualizácia všetkých piatich Lagrangeových bodov
body_L45 = points([L4, L5], color='lime', marker='x', size=100, zorder=10, legend_label='L4, L5')

vysledna_mapa_vsetky = vysledna_mapa_E + body_L123 + body_L45

vysledna_mapa_vsetky.show(
    frame=True,
    title="Všetkých päť Lagrangeových bodov systému Zem-Mesiac",
    axes_labels=["$x$", "$y$"],
    figsize=10
)
```
